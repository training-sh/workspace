## Phase 1: move the first day
Use a rewrite here. The Hive source and Iceberg destination own separate files, which makes coexistence and rollback easier.
The migration boundary is a business date. A record belongs to exactly one side of the combined read view.
This example has no concurrent writer; in production record a durable watermark and capture changes during backfill.


```python
# Create the table for this step.
spark.sql(f"""
    CREATE TABLE polaris.{ns}.orders
    USING iceberg
    PARTITIONED BY (days(order_ts))
    TBLPROPERTIES ('format-version'='2')
    AS SELECT *
    FROM spark_catalog.{ns}.orders
    WHERE order_date < DATE '2025-01-02'
""").show(truncate=False)
```

```python
# Expose this SQL result as a temporary view.
spark.sql(f"""
    CREATE OR REPLACE TEMP VIEW orders_live AS
    SELECT *
    FROM polaris.{ns}.orders
    WHERE order_date < DATE '2025-01-02'
    UNION ALL
    SELECT *
    FROM spark_catalog.{ns}.orders
    WHERE order_date >= DATE '2025-01-02'
""")
```

```python
# No unmatched rows should remain in this direction.
assert spark.sql(f"""
    (SELECT *
    FROM spark_catalog.{ns}.orders) EXCEPT ALL (SELECT *
    FROM orders_live)
""").count() == 0, 'No unmatched rows should remain in this direction'
```

```python
# No unmatched rows should remain in this direction.
assert spark.sql(f"""
    (SELECT *
    FROM orders_live) EXCEPT ALL (SELECT *
    FROM spark_catalog.{ns}.orders)
""").count() == 0, 'No unmatched rows should remain in this direction'

print('PASS: complete row reconciliation')
```

## Phase 2: backfill another partition
Keep the old view boundary while loading the second day. Then validate the backfill and advance the boundary.
Using a key-based MERGE makes this backfill repeatable. The source must have one row per business key.


```python
# Keep this value for the next comparison.
backfill_sql = f"""
    MERGE INTO polaris.{ns}.orders t
    USING (SELECT *
    FROM spark_catalog.{ns}.orders
    WHERE order_date = DATE '2025-01-02') s
    ON t.order_id = s.order_id
    WHEN NOT MATCHED THEN INSERT *
"""
```

```python
# Run this SQL statement and inspect its result.
spark.sql(backfill_sql)
```

```python
# Run this SQL statement and inspect its result.
spark.sql(backfill_sql)
```

```python
# Backfill replay did not duplicate rows.
assert spark.table(f'polaris.{ns}.orders').count() == 4, 'Backfill replay did not duplicate rows'

print('PASS: Backfill replay did not duplicate rows')
```

```python
# Old read boundary remains complete during backfill.
assert spark.table('orders_live').count() == 6, 'Old read boundary remains complete during backfill'

print('PASS: Old read boundary remains complete during backfill')
```

```python
# Expose this SQL result as a temporary view.
spark.sql(f"""
    CREATE OR REPLACE TEMP VIEW orders_live AS
    SELECT *
    FROM polaris.{ns}.orders
    WHERE order_date < DATE '2025-01-03'
    UNION ALL
    SELECT *
    FROM spark_catalog.{ns}.orders
    WHERE order_date >= DATE '2025-01-03'
""")
```

```python
# No unmatched rows should remain in this direction.
assert spark.sql(f"""
    (SELECT *
    FROM spark_catalog.{ns}.orders) EXCEPT ALL (SELECT *
    FROM orders_live)
""").count() == 0, 'No unmatched rows should remain in this direction'
```

```python
# No unmatched rows should remain in this direction.
assert spark.sql(f"""
    (SELECT *
    FROM orders_live) EXCEPT ALL (SELECT *
    FROM spark_catalog.{ns}.orders)
""").count() == 0, 'No unmatched rows should remain in this direction'

print('PASS: complete row reconciliation')
```

## Prove why an unbounded union is wrong
Historical rows now exist in both stores. The following count reveals the overlap without changing any data.


```python
# Read the rows or metadata needed for this check.
wrong_count = spark.sql(f"""
    SELECT *
    FROM polaris.{ns}.orders
    UNION ALL SELECT *
    FROM spark_catalog.{ns}.orders
""").count()
```

```python
# Unbounded union double-counts four migrated records.
assert wrong_count == 10, 'Unbounded union double-counts four migrated records'

print('PASS: Unbounded union double-counts four migrated records')
```

```python
# Bounded view has exactly six records.
assert spark.table('orders_live').count() == 6, 'Bounded view has exactly six records'

print('PASS: Bounded view has exactly six records')
```

## Phase 3: freeze, catch up, reconcile and switch readers
At this point source writes must be frozen. Capture and apply the final changes before changing the read path.
A temporary view makes the transition visible in this lab. A real deployment needs a governed view or an application configuration change.


```python
# Apply the source rows by business key.
spark.sql(f"""
    MERGE INTO polaris.{ns}.orders t
    USING spark_catalog.{ns}.orders s
    ON t.order_id = s.order_id
    WHEN MATCHED AND s.updated_at > t.updated_at THEN UPDATE SET *
    WHEN NOT MATCHED THEN INSERT *
""")
```

```python
# No unmatched rows should remain in this direction.
assert spark.sql(f"""
    (SELECT *
    FROM spark_catalog.{ns}.orders) EXCEPT ALL (SELECT *
    FROM polaris.{ns}.orders)
""").count() == 0, 'No unmatched rows should remain in this direction'
```

```python
# No unmatched rows should remain in this direction.
assert spark.sql(f"""
    (SELECT *
    FROM polaris.{ns}.orders) EXCEPT ALL (SELECT *
    FROM spark_catalog.{ns}.orders)
""").count() == 0, 'No unmatched rows should remain in this direction'

print('PASS: complete row reconciliation')
```

```python
# Expose this SQL result as a temporary view.
spark.sql(f"""
    CREATE OR REPLACE TEMP VIEW orders_live
    AS SELECT *
    FROM polaris.{ns}.orders
""")
```

```python
# Final table has independent S3 files.
assert all((p.startswith('s3://') for p in {file_row.file_path for file_row in spark.sql(f"""
    SELECT file_path
    FROM polaris.{ns}.orders.files
""").collect()})), 'Final table has independent S3 files'

print('PASS: Final table has independent S3 files')
```

```python
# Read path is now entirely Iceberg.
assert spark.table('orders_live').count() == 6, 'Read path is now entirely Iceberg'

print('PASS: Read path is now entirely Iceberg')
```
