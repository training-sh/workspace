## Stage the source data in HDFS
These six orders span three days. Amounts use fixed precision and all timestamps use UTC.
The explicit `order_date` column represents a traditional Hive partition key.


```python
# Define the staged rows and their explicit column types.
staged_orders_df = spark.createDataFrame(
    [
        (101, 1, datetime(2025, 1, 1, 9), Decimal('120.00'), 'PLACED', datetime(2025, 1, 1, 9), date(2025, 1, 1)),
        (102, 2, datetime(2025, 1, 1, 11), Decimal('80.00'), 'PLACED', datetime(2025, 1, 1, 11),
            date(2025, 1, 1)),
        (103, 1, datetime(2025, 1, 2, 10), Decimal('200.00'), 'PAID', datetime(2025, 1, 2, 12), date(2025, 1, 2)),
        (104, 3, datetime(2025, 1, 2, 15), Decimal('50.00'), 'PLACED', datetime(2025, 1, 2, 15),
            date(2025, 1, 2)),
        (105, 2, datetime(2025, 1, 3, 8), Decimal('300.00'), 'PAID', datetime(2025, 1, 3, 10), date(2025, 1, 3)),
        (106, 4, datetime(2025, 1, 3, 18), Decimal('150.00'), 'PLACED', datetime(2025, 1, 3, 18),
            date(2025, 1, 3)),
    ],
    """
        order_id INT,
        customer_id INT,
        order_ts TIMESTAMP,
        amount DECIMAL(12,2),
        status STRING,
        updated_at TIMESTAMP,
        order_date DATE
    """,
)
```

## Catch type problems before file import
Do not import a string amount as a decimal just because the column has the same name.
Write the raw feed to HDFS, normalize it with SQL, and separate invalid rows before migration.
`try_cast` makes bad input visible as null instead of failing halfway through a load.
An explicit timestamp pattern avoids treating a local clock value as UTC by accident.


```python
# Define the staged rows and their explicit column types.
staged_invoices_df = spark.createDataFrame(
    [
        ('501', '120.00', '2025-01-01T09:00:00Z'),
        ('502', 'bad-amount', '2025-01-01T11:00:00Z'),
        ('503', '200.00', 'bad-time'),
        ('504', '99999999999999.99', '2025-01-02T10:00:00Z'),
    ],
    """
        invoice_id STRING,
        amount STRING,
        invoice_ts STRING
    """,
)
```

```python
# Keep the raw invoice feed separate from normalized data.
raw_path = root + '/invoices/raw-strings'

staged_invoices_df.write.mode('errorifexists').parquet(raw_path)
```

```python
# Make these rows available to Spark SQL.
spark.read.parquet(raw_path).createOrReplaceTempView('raw_invoices')
```

```python
# Expose this SQL result as a temporary view.
spark.sql("""
    CREATE OR REPLACE TEMP VIEW normalized_invoices AS
    SELECT try_cast(invoice_id AS INT) AS invoice_id,
    try_cast(amount AS DECIMAL(12,2)) AS amount,
    try_cast(invoice_ts AS TIMESTAMP) AS invoice_ts
    FROM raw_invoices
""")
```

```python
# Expose this SQL result as a temporary view.
spark.sql("""
    CREATE OR REPLACE TEMP VIEW rejected_invoices AS
    SELECT *
    FROM normalized_invoices
    WHERE invoice_id IS NULL OR amount IS NULL OR invoice_ts IS NULL
""")
```

```python
# Expose this SQL result as a temporary view.
spark.sql("""
    CREATE OR REPLACE TEMP VIEW accepted_invoices AS
    SELECT *
    FROM normalized_invoices
    WHERE invoice_id IS NOT NULL AND amount IS NOT NULL AND invoice_ts IS NOT NULL
""")
```

```python
# Read the rows or metadata needed for this check.
spark.sql("""
    SELECT *
    FROM rejected_invoices
""").show(truncate=False)
```

```python
# Bad amount, bad timestamp and decimal overflow are quarantined.
assert spark.table('rejected_invoices').count() == 3, 'Bad amount, bad timestamp and decimal overflow are quarantined'

print('PASS: Bad amount, bad timestamp and decimal overflow are quarantined')
```

```python
# One invoice is safe to migrate.
assert spark.table('accepted_invoices').count() == 1, 'One invoice is safe to migrate'

print('PASS: One invoice is safe to migrate')
```

```python
# Create the table for this step.
spark.sql(f"""
    CREATE TABLE polaris.{ns}.invoices_clean
    USING iceberg
    AS SELECT *
    FROM accepted_invoices
""")
```

```python
# No unmatched rows should remain in this direction.
assert spark.sql(f"""
    (SELECT *
    FROM accepted_invoices) EXCEPT ALL (SELECT *
    FROM polaris.{ns}.invoices_clean)
""").count() == 0, 'No unmatched rows should remain in this direction'
```

```python
# No unmatched rows should remain in this direction.
assert spark.sql(f"""
    (SELECT *
    FROM polaris.{ns}.invoices_clean) EXCEPT ALL (SELECT *
    FROM accepted_invoices)
""").count() == 0, 'No unmatched rows should remain in this direction'

print('PASS: complete row reconciliation')
```

## Reconciliation includes keys and totals
Duplicate business keys can survive a row-count check. Check uniqueness separately, and compare all rows.


```python
# Read the rows or metadata needed for this check.
spark.sql(f"""
    SELECT order_date, count(*) AS orders, sum(amount) AS amount
    FROM spark_catalog.{ns}.orders
    GROUP BY order_date
    ORDER BY order_date
""").show(truncate=False)
```

```python
# Source business keys are unique.
assert spark.sql(f"""
    SELECT order_id
    FROM spark_catalog.{ns}.orders
    GROUP BY order_id
    HAVING count(*) > 1
""").count() == 0, 'Source business keys are unique'

print('PASS: Source business keys are unique')
```

```python
# Remove only the specified catalog object.
spark.sql(f"""
    DROP TABLE spark_catalog.{ns}.orders_trial
""")
```

```python
# Dropping the test registration retained all source files.
assert all((hdfs.exists(HadoopPath(p)) for p in source_files)), 'Dropping the test registration retained all source files'

print('PASS: Dropping the test registration retained all source files')
```

```python
# Source remains readable after the test.
assert spark.table(f'spark_catalog.{ns}.orders').count() == 6, 'Source remains readable after the test'

print('PASS: Source remains readable after the test')
```
