## Traditional Hive partitions
Hive stores `order_date` in partition metadata and directory names. A timestamp filter alone does not state a partition-key filter.
The explicit date predicate below is useful only if the application keeps it consistent with `order_ts`.


```python
# Inspect how Spark plans this query.
spark.sql(f"""
    EXPLAIN FORMATTED SELECT order_id, amount
    FROM spark_catalog.{ns}.orders
    WHERE order_date = DATE '2025-01-02'
    AND order_ts >= TIMESTAMP '2025-01-02 00:00:00' AND order_ts < TIMESTAMP '2025-01-03 00:00:00'
""").show(truncate=False)
```

## Iceberg derives the partition value
The Iceberg schema omits `order_date`. Its partition spec derives a day from `order_ts`.
Applications filter the timestamp. Iceberg translates that predicate into partition pruning and file-level filtering.
Hidden partitioning does not mean the metadata is secret; inspect it through metadata tables.


```python
# Create the table for this step.
spark.sql(f"""
    CREATE TABLE polaris.{ns}.orders
    USING iceberg
    PARTITIONED BY (days(order_ts))
    TBLPROPERTIES ('format-version'='2')
    AS SELECT order_id, customer_id, order_ts, amount, status, updated_at
    FROM spark_catalog.{ns}.orders
""").show(truncate=False)
```

```python
# No explicit partition column in the business schema.
assert 'order_date' not in spark.table(f'polaris.{ns}.orders').columns, 'No explicit partition column in the business schema'

print('PASS: No explicit partition column in the business schema')
```

```python
# Keep this value for the next comparison.
day_query = f"""
    SELECT order_id, amount
    FROM polaris.{ns}.orders
    WHERE order_ts >= TIMESTAMP '2025-01-02 00:00:00'
    AND order_ts < TIMESTAMP '2025-01-03 00:00:00'
"""
```

```python
# Timestamp-only query returns the two expected orders.
assert spark.sql(day_query).count() == 2, 'Timestamp-only query returns the two expected orders'

print('PASS: Timestamp-only query returns the two expected orders')
```

```python
# Run this SQL statement and inspect its result.
spark.sql('EXPLAIN FORMATTED ' + day_query).show(truncate=False)
```

```python
# Read the rows or metadata needed for this check.
spark.sql(f"""
    SELECT spec_id, partition, record_count, file_path
    FROM polaris.{ns}.orders.files
""").show(truncate=False)
```

```python
# Read the rows or metadata needed for this check.
original_paths = {file_row.file_path for file_row in spark.sql(f"""
    SELECT file_path
    FROM polaris.{ns}.orders.files
""").collect()}
```

## Evolve the specification
Change days to months and add a customer bucket. Existing files keep their old specification;
only new writes use the new one. Bucket transforms are hashes, not customer ranges or Hive bucket directories.
Eight buckets demonstrate syntax here; choose production counts from volume, skew and workload evidence.


```python
# Apply the table metadata change.
spark.sql(f"""
    ALTER TABLE polaris.{ns}.orders REPLACE PARTITION FIELD days(order_ts) WITH months(order_ts)
""")
```

```python
# Apply the table metadata change.
spark.sql(f"""
    ALTER TABLE polaris.{ns}.orders ADD PARTITION FIELD bucket(8, customer_id)
""")
```

```python
# Changing the spec did not rewrite existing files.
assert {file_row.file_path for file_row in spark.sql(f"""
    SELECT file_path
    FROM polaris.{ns}.orders.files
""").collect()} == original_paths, 'Changing the spec did not rewrite existing files'

print('PASS: Changing the spec did not rewrite existing files')
```

```python
# Define the staged rows and their explicit column types.
staged_new_orders_df = spark.createDataFrame(
    [
        (107, 5, datetime(2025, 2, 1, 10), Decimal('75.00'), 'PLACED', datetime(2025, 2, 1, 10)),
    ],
    """
        order_id INT,
        customer_id INT,
        order_ts TIMESTAMP,
        amount DECIMAL(12,2),
        status STRING,
        updated_at TIMESTAMP
    """,
)
```

```python
# Keep the new-specification input in a separate HDFS directory.
new_path = root + '/orders/new-spec'

staged_new_orders_df.write.mode('errorifexists').parquet(new_path)
```

```python
# Make these rows available to Spark SQL.
spark.read.parquet(new_path).createOrReplaceTempView('new_orders')
```

```python
# Write this batch into the destination table.
spark.sql(f"""
    INSERT INTO polaris.{ns}.orders SELECT *
    FROM new_orders
""")
```

```python
# Old and new specifications coexist.
assert spark.sql(f"""
    SELECT DISTINCT spec_id
    FROM polaris.{ns}.orders.files
""").count() == 2, 'Old and new specifications coexist'

print('PASS: Old and new specifications coexist')
```

```python
# Business query works across specification changes.
assert spark.sql(day_query).count() == 2, 'Business query works across specification changes'

print('PASS: Business query works across specification changes')
```

## Rewrite existing files into the current specification
Specification changes are cheap metadata operations. Rewriting old data has a separate compute and I/O cost.
Request a full rewrite explicitly in this small lab, then reconcile all seven rows.


```python
# Run the Iceberg procedure for this step.
spark.sql(f"""
    CALL polaris.system.rewrite_data_files(
    table => '{ns}.orders', options => map('rewrite-all', 'true')
    )
""").show(truncate=False)
```

```python
# Active files use only the current specification.
assert spark.sql(f"""
    SELECT DISTINCT spec_id
    FROM polaris.{ns}.orders.files
""").count() == 1, 'Active files use only the current specification'

print('PASS: Active files use only the current specification')
```

```python
# No unmatched rows should remain in this direction.
assert spark.sql(f"""
    (SELECT order_id, customer_id, order_ts, amount, status, updated_at
    FROM staged_orders
    UNION ALL SELECT *
    FROM new_orders) EXCEPT ALL (SELECT *
    FROM polaris.{ns}.orders)
""").count() == 0, 'No unmatched rows should remain in this direction'
```

```python
# No unmatched rows should remain in this direction.
assert spark.sql(f"""
    (SELECT *
    FROM polaris.{ns}.orders) EXCEPT ALL (SELECT order_id, customer_id, order_ts, amount, status, updated_at
    FROM staged_orders
    UNION ALL SELECT *
    FROM new_orders)
""").count() == 0, 'No unmatched rows should remain in this direction'

print('PASS: complete row reconciliation')
```
